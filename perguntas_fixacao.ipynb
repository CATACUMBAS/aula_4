1. **(a) Sistema de previsão de demanda que roda 1 vez por dia
Parcialmente observável — não vê todas as variáveis reais que influenciam a demanda.**

Estocástico — demanda envolve incerteza.

Episódico — cada previsão diária é independente.

Estático — o ambiente não muda enquanto o sistema calcula.

Contínuo — valores de demanda são numéricos.

Agente único — não interage com outros agentes.

**(b) Drone de inspeção de linhas de transmissão
Parcialmente observável**

Estocástico — vento, clima, falhas.

Sequencial — decisões afetam estados futuros.

Dinâmico — ambiente muda enquanto o drone age.

Contínuo — posição, velocidade, sinais.

Agente único — normalmente opera sozinho.

**(c) Bot que negocia ações na bolsa
Parcialmente observável**

Estocástico — mercado é imprevisível.

Sequencial — ações passadas influenciam resultados futuros.

Dinâmico — mercado muda constantemente.

Contínuo — preços variam continuamente.

Multiagente — interage com milhares de outros agentes.

**(d) Corretor ortográfico
Totalmente observável**

Determinístico — correção segue regras fixas.

Episódico — cada palavra é tratada separadamente.

Estático — texto não muda sozinho.

Discreto — palavras e símbolos são discretos.

Agente único — não interage com outros agentes.

3. Agente com percepção parcial não pode ser perfeitamente racional -> **falso**  
racionalidade depende de maximizar desempenho esperado, não de ter percepção completa  

Existe ambiente em que todo agente é racional -> **verdadeiro**  
Ambiente determinístico, totalmente observável, estático e com medida de desempenho trivial.

A medida de desempenho sempre pode ser escrita como função da sequência de percepções -> **verdadeiro**

AGENTE BASEADO EM UTILIDADE

```
fun agenteUtilidade(percepcao):
    acoes = [ASPIRAR, MOVER_ESQ, MOVER_DIR, ESPERAR]
    melhorAcao = ESPERAR
    melhorUtilidade = -infinito

    for acao in acoes:
        util = calcularUtilidade(acao, percepcao)
        if util > melhorUtilidade:
            melhorUtilidade = util
            melhorAcao = acao

    return melhorAcao
```


5. Por que agente baseado em tabelas é inviável?
Se o agente recebe 10 bits por segundo durante 1 hora:

percepcoes por hora 3600s X 10 bits = 36000 bits

Número de sequências possíveis:
2
36000

Esse número é absurdamente grande — maior que o número de átomos no universo.
Portanto, impossível armazenar ou construir.
